문제6. 아래 설명 중 틀린 내용을 모두 골라주세요

1. CommandNames 들은 쉘의 검색 PATH 안에 있어야 합니다.
2. 쉘은 터미널을 통해 명령이 제출될 때 명령의 의미를 해석하는 것입니다.
3. Linux 명령은 CommandName -options input 구조를 따릅니다.
4. 일부 명령에는 짧은 형식 옵션만 사용할 수 있습니다.
5. 긴 형식 옵션에는 대시 1개 (-)가 있고 짧은 형식 옵션에는 대시 2개(--)가 있습니다.

````python
#!/bin/bash

echo "문제 6: 틀린 설명을 확인합니다."

# 각 설명의 정오 여부를 확인
wrong_statements=()

# 1번 검증: CommandNames는 반드시 PATH 안에 있을 필요는 없음 (절대경로, alias, 함수 가능)
if true; then
    wrong_statements+=("1")
fi

# 2번 검증: 쉘은 단순한 터미널 입력 해석이 아니라, 스크립트 실행, 변수 처리 등도 수행
if true; then
    wrong_statements+=("2")
fi

# 3번 검증: 일반적인 형식이지만 반드시 "-options" 형태가 필요한 것은 아님 (옵션 없이 실행되는 명령어도 많음)
if true; then
    wrong_statements+=("3")
fi

# 4번 검증: 일부 명령어는 긴 옵션만 존재하거나, 짧고 긴 옵션을 둘 다 제공할 수 있음
if true; then
    wrong_statements+=("4")
fi

# 5번 검증: 긴 옵션은 `--`, 짧은 옵션은 `-`이 맞음 → 해당 설명은 맞음
if false; then
    wrong_statements+=("5")
fi

# 결과 출력
if [ ${#wrong_statements[@]} -eq 0 ]; then
    echo "틀린 설명이 없습니다."
else
    echo "틀린 설명: ${wrong_statements[*]}"
fi
````
